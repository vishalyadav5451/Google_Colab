```python
house_data = {
    "area": [800, 900, 1200, 1500, 700, 850, 1100, 1300, 600, 750, 1000, 1400, 900, 1150, 1600],
    "bedrooms": [2, 2, 3, 3, 1, 2, 3, 3, 1, 2, 2, 3, 2, 3, 4],
    "bathrooms": [1, 2, 2, 3, 1, 1, 2, 2, 1, 1, 2, 3, 2, 2, 3],
    "location": [
        "Jaipur", "Jaipur", "Jaipur", "Jaipur",
        "Ajmer", "Ajmer", "Ajmer", "Ajmer",
        "Kota", "Kota", "Kota", "Kota",
        "Udaipur", "Udaipur", "Udaipur"  ],
    "price": [
        3200000, 3800000, 5200000, 6500000,
        2500000, 3000000, 4800000, 5500000,
        2200000, 2800000, 4200000, 6000000,
        4000000, 5100000, 7200000
    ]
}
```

```python
import numpy as np
import pandas as pd
df = pd.DataFrame(house_data)
```

```python
df.info()
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
le = LabelEncoder()
df['location'] = le.fit_transform(df['location'])
```

```python
df
```

```python
x = df.drop('price' , axis =1)
y = df['price']
```

```python
from sklearn.model_selection import train_test_split
x_train , x_test , y_train , y_test = train_test_split(x,y , test_size=0.2 , random_state = 42 )


```

```python
from sklearn.preprocessing import StandardScaler
ss = StandardScaler()

X_train = ss.fit_transform(x_train)
X_test = ss.transform(x_test)
```

```python
from sklearn.linear_model import LinearRegression
model =LinearRegression()
```

```python
model.fit(x_train , y_train)
```

```python
pred = model.predict(x_test)
```

```python
pred
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
print(mean_absolute_error(y_test , pred))
print(mean_squared_error(y_test , pred))
print(r2_score(y_test , pred))
```

```python
print("RMSE:", np.sqrt(mean_squared_error(y_test,pred)))
```

```python
from sklearn.ensemble import RandomForestRegressor
```

```python
rf = RandomForestRegressor(n_estimators=100 , random_state = 42 )
```

```python
rf.fit(x_train , y_train)
```

```python
rf_pred=rf.predict(x_test)
```

```python
r2_score(y_test , rf_pred)
```
